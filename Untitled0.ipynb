# Titanic Dataset: Data Cleaning and Preprocessing

## 1. Introduction

The purpose of this project is to perform data cleaning and preprocessing on the Titanic passenger dataset obtained from Kaggle. The dataset contains passenger information, including age, gender, passenger class, ticket fare, and survival status.

Real-world datasets can contain missing values, duplicate records, inconsistent categories, and numerical features with different scales. These problems can affect data analysis and machine-learning models.

In this project, I will inspect the dataset, identify and handle data-quality problems, create useful features, encode categorical variables, transform skewed numerical features, and standardize numerical data where appropriate.

The target variable is `Survived`, where 0 represents a passenger who did not survive and 1 represents a passenger who survived.

```python

print("Duplicate rows before cleaning:", df.duplicated().sum())

df = df.drop_duplicates().copy()

print("Duplicate rows after cleaning:", df.duplicated().sum())
print("Shape after duplicate removal:", df.shape)
```

```python

# Make a working copy
df = df.copy()

# Standardize text formatting
for col in ["Sex", "Embarked"]:
    df[col] = df[col].apply(
        lambda x: x.strip() if isinstance(x, str) else x
    )

# Standardize category capitalization
df["Sex"] = df["Sex"].str.lower()
df["Embarked"] = df["Embarked"].str.upper()

# Check allowed category values
print("Sex categories:", df["Sex"].unique())
print("Embarked categories:", df["Embarked"].unique())

# Correct numerical data types
for col in ["PassengerId", "Survived", "Pclass",
            "SibSp", "Parch"]:
    df[col] = pd.to_numeric(df[col], errors="coerce").astype("Int64")

for col in ["Age", "Fare"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")

print("\nUpdated data types:")
print(df.dtypes)
```

```python

for col in ["Age", "Fare", "SibSp", "Parch"]:
    values = df[col].dropna()

    Q1 = values.quantile(0.25)
    Q3 = values.quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = values[(values < lower) | (values > upper)]

    print(f"{col}: {len(outliers)} potential outliers")
    print(f"Lower bound: {lower:.2f}, Upper bound: {upper:.2f}")
```

```python

# Whether cabin information exists
df["HasCabin"] = df["Cabin"].notna().astype(int)

# Total family members travelling with the passenger, including the passenger
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1

# Whether the passenger travelled alone
df["IsAlone"] = (df["FamilySize"] == 1).astype(int)

# Transform the fare using log1p to reduce right skew
df["Fare_log"] = np.log1p(df["Fare"])

print("New features:")
display(df[["FamilySize", "IsAlone", "HasCabin", "Fare", "Fare_log"]].head())

print("\nFare skewness before transformation:", df["Fare"].skew())
print("Fare skewness after transformation:", df["Fare_log"].skew())
```

```python

from sklearn.model_selection import train_test_split

# Target variable
y = df["Survived"].astype(int)

# Remove target and columns not used in the basic model
X = df.drop(
    columns=[
        "Survived", "PassengerId", "Name",
        "Ticket", "Cabin", "Fare"
    ]
)

# Split before fitting imputation, encoding, or scaling
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)
print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
```

```python

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# Numerical features
numeric_features = [
    "Age", "SibSp", "Parch", "FamilySize",
    "IsAlone", "HasCabin", "Pclass", "Fare_log"
]

# Categorical features
categorical_features = ["Sex", "Embarked"]

# Numerical preprocessing:
# Fill missing values with median, then standardize
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Categorical preprocessing:
# Fill missing values with most frequent category,
# then apply one-hot encoding
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# Combine both preprocessing pipelines
preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_features),
    ("categorical", categorical_pipeline, categorical_features)
])

# Learn preprocessing from training data only
X_train_processed = preprocessor.fit_transform(X_train)

# Apply the learned preprocessing to test data
X_test_processed = preprocessor.transform(X_test)

print("Processed training shape:", X_train_processed.shape)
print("Processed testing shape:", X_test_processed.shape)

print("\nRemaining missing values in processed training data:",
      np.isnan(X_train_processed.toarray()
               if hasattr(X_train_processed, "toarray")
               else X_train_processed).sum())
```
